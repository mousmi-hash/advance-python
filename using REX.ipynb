```python
#3. Reverse a string (without slicing)
s = input()
rev = ""
for ch in s:
    rev = ch + rev
print(rev)
```

```python
#4. Count vowels using regex
import re
s = input()
vowels = re.findall(r"[aeiouAEIOU]", s)
print(len(vowels))
```

```python
#5. Sum of digits
n = input()
print(sum(int(d) for d in n))
```

```python
#6. Palindrome check
s = input()
print("Palindrome" if s == s[::-1] else "Not Palindrome")
```

```python
#7. Fibonacci up to N terms
n = int(input())
a, b = 0, 1
for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#8. Swap without third variable
a, b = map(int, input().split())
a, b = b, a
print(a, b)
```

```python
#9. Count words
import re
s = input()
words = re.findall(r"\b\w+\b", s)
print(len(words))
```

```python
#11. Second largest number
lst = list(map(int, input().split()))
lst = list(set(lst))
lst.sort()
print(lst[-2])
```

```python
#12. Remove duplicates
lst = list(map(int, input().split()))
res = []
for i in lst:
    if i not in res:
        res.append(i)
print(res)
```

```python
#13. Frequency of elements
lst = list(map(int, input().split()))
freq = {}
for i in lst:
    freq[i] = freq.get(i, 0) + 1
print(freq)
```

```python
#16. Rotate list by k positions
lst = list(map(int, input().split()))
k = int(input())
k = k % len(lst)
print(lst[k:] + lst[:k])
```

```python
#17. Anagram check
import re
s1 = re.sub(r"\W", "", input().lower())
s2 = re.sub(r"\W", "", input().lower())

print("Anagram" if sorted(s1) == sorted(s2) else "Not Anagram")
```

```python
#18. First Non-Repeating Character
s = input("Enter string: ")

for ch in s:
    if s.count(ch) == 1:
        print(ch)
        break
```

```python
#19. Flatten Nested List
lst = [[1,2],[3,4],[5]]

flat = []
for i in lst:
    flat.extend(i)

print(flat)
```

```python
#20. Pairs with Given Sum
lst = list(map(int, input("Enter list: ").split()))
target = int(input("Enter target: "))

seen = set()
for num in lst:
    if target - num in seen:
        print((num, target - num))
    seen.add(num)
```

```python
#21. Right Triangle Pattern
n = int(input("Enter number: "))

for i in range(1, n+1):
    print("*" * i)
```

```python
#22. Pyramid Pattern
n = int(input("Enter number: "))

for i in range(1, n+1):
    print(" "*(n-i) + "*"*(2*i-1))
```

```python
#23. Prime Numbers in Range
a, b = map(int, input("Enter range: ").split())

for num in range(a, b+1):
    if num > 1:
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

```python
#24. Armstrong Number
n = input("Enter number: ")

power = len(n)
total = sum(int(d)**power for d in n)

print("Armstrong" if total == int(n) else "Not Armstrong")
```
